```python
'''
Đoạn này thử đưa một đoạn dữ liệu vào để load. Sau khi chạy xong thì dữ liệu raw sẽ được lưu trong 'test_data'. Người dùng có thể thực hiện filter hoặc tính tổng từ đây.
'''

test_path = r'..\data\raw\data_Duoc_full\raw\kqtt\202504-202505.xls' # đưa PATH đến data ở đây. 
test_data = LoadMedicalData(
    test_path, 
    config=producer_config # Phần config này về sau sẽ cần mở rộng cho các dạng filter khác, hiện tại đang mới có filter dành cho 'nhasx'
)
```

```python
# Hiển thị toàn bộ dữ liệu raw
test_data.show_data()
```

```python
# Filter dữ liệu tùy theo config đưa vào
test_data.filter().show_data() # Nếu để trống phần filter sẽ trả về toàn bộ các mục mà được liệt kê trong config
```

```python
# Filter ra một mục tùy chọn
test_data.filter(by='IMP').show_data() 
```

```python
# Tính tổng số lượng
print(f'Tổng số lượng của IMP: {test_data.filter(by='IMP').sum(value_col='soluong'):,.0f}') # Cái ':,.0f' là để hiển thị dấu phẩy hàng nghìn
 
# Tính tổng giá trị
print(f'Tổng giá trị của IMP: {test_data.filter(by='IMP').sum(value_col='thanhtien'):,.0f}')
```
