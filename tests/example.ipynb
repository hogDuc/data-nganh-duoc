# Lọc (filter) dữ liệu

_Lưu ý:_ Cần sử dụng .show_data() ở sau các bước để hiện dữ liệu

```python
# Filter theo nhà sản xuất
loader.filter(column='ticker', value='IMP').show_data()
```

```python
# Filter theo quốc gia sản xuất
loader.filter(column='country', value='Hoa Kỳ').show_data()
```

```python
# Filter theo nhóm thầu
loader.filter(column='nhomthau', value='N1').show_data()
```

```python
# Filter nhiều bước
loader.filter(
    column='Loại thuốc 1', 
    value='6. Thuốc điều trị ký sinh trùng, chống nhiễm khuẩn'
).filter(
    column='country',
    value='Hoa Kỳ'
).show_data()
```

# Tính tổng, giá trị trung bình

```python
# Tính tổng số lượng theo quốc gia
loader.average(
    value_col='soluong',
    groupby='country'
)
```

```python
# Tính tổng số lượng theo quốc gia trong khoảng thời gian công bố nhất định
loader.filter(
    column='congbo',
    value=pd.to_datetime('2026-04-24')
).average(
    value_col='soluong',
    groupby='country'
)
```

```python
# Tính giá trung bình theo nhóm
loader.average(
    value_col='gia',
    groupby='Loại thuốc 2'
)
```
